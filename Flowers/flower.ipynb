```python
from CPR_plotly.Arcs.arc import *
```

```python
import plotly.graph_objects as go

fig = go.Figure()
fig.update_layout(
    # width=800, height=600,
    autosize=True  # 像素宽度  # 像素高度  # 关闭自动调整
)
fig.update_layout(
    title={
        "text": "自定义标题",
        "font": {"size": 24, "color": "darkblue"},
        "x": 0.5,  # 居中标题
    },
    margin=dict(l=50, r=50, t=80, b=50),  # 边距设置
    paper_bgcolor="#303",  # 画布背景色
    plot_bgcolor="#eee",  # 绘图区背景色
    xaxis=dict(showgrid=True, gridwidth=1, gridcolor="#00f"),
)
fig.update_layout(autosize=True, hovermode="closest", dragmode="pan")
```

```python
arc((0,0),(1,0),(0,1))
```

```python
# 导出为静态图像 
fig.write_image( 
    "custom_plot.png", 
    scale=2,          # 2倍高清输出 
    engine="kaleido"  # 推荐渲染引擎 
)
```

```python
# HTML交互式输出 
fig.write_html( 
    "plot.html", 
    full_html=False,
    include_plotlyjs="cdn"  # CDN加载JS库 
)
```
